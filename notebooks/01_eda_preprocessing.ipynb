**Using the "Store Closed" Insight for Data Cleaning**

- Incorporating this domain rule, (Store closed, sales = 0) is one of the most impactful data cleaning steps in a retail supply chain project.

- The Problem: Raw retail datasets often contain records where a store was closed (due to holidays, renovations, or unexpected shutdowns), but some rows might have missing values, NaN markers, or noisy imputed numbers in the sales column. If left unaddressed, models try to learn false demand patterns from closed days.

- The Solution: Applying the business logic rule— **"When a store is closed (Open == 0), sales must equal 0"**—forces absolute data integrity.
- This rule is executed in Notebook 1(Step 2.2), using a direct pandas boolean mask:
    Domain-Driven Data Cleaning Rule: When a store is closed, sales = 0.

   Code: **train_df.loc[train_df['Open'] == 0, 'Sales'] = 0**

- Benefits of This Step
    - Elimination of Artificial Noise: Prevents machine learning models from wasting learning capacity trying to predict sales when a store's doors are physically locked.

    - Reduction of Error Metrics: Dramatically cuts down large residual errors during evaluation, ensuring your model doesn't get penalized for predicting high sales on a day the store was closed.

    - Supply Chain Alignment: Reflects true business reality. A closed store generates zero revenue; hardcoding this eliminates inventory distortion and helps supply chain managers calculate precise safety stock requirements.

```python
# =============================================================================
# 2.5 : Feature Engineering | Date-Time Transformations & One-Hot Encoding
# =============================================================================

# Extract temporal components safely from datetime objects for both train and test sets
for df in [train_df, test_df]:
    if 'Year' not in df.columns:
        df['Year'] = df['Date'].dt.year
        df['Month'] = df['Date'].dt.month
        df['Day'] = df['Date'].dt.day
        # Use .fillna(0).astype(int) to safely handle any missing calendar week values
        df['WeekOfYear'] = df['Date'].dt.isocalendar().week.fillna(0).astype(int)

# One-Hot Encoding for StateHoliday (handling categorical values only if the column exists)
if 'StateHoliday' in train_df.columns:
    train_df = pd.get_dummies(train_df, columns=['StateHoliday'], prefix='StateHoliday', drop_first=True)
if 'StateHoliday' in test_df.columns:
    test_df = pd.get_dummies(test_df, columns=['StateHoliday'], prefix='StateHoliday', drop_first=True)

# Align columns between train and test datasets after dummy variable generation
train_df, test_df = train_df.align(test_df, join='left', axis=1, fill_value=0)

print(f"[Feature Engineering] Train Shape after transformations: {train_df.shape}")
print(f"[Feature Engineering] Test Shape after transformations: {test_df.shape}")
```

```python
#===============================================================================
# 2.6 : Data Preprocessing | Outliers Inspection, Detection, and Feature Scaling
#===============================================================================

print("-" * 80  )
print("[Data Preprocessing] Outliers Inspection, Detection, and Feature Scaling")
print("-" * 80  )

train_df_num = train_df.select_dtypes(include=[np.number]).columns.tolist()

skewness_records = []
for col in train_df_num:
    clean_series = pd.to_numeric(train_df[col], errors='coerce').dropna()
    if len(clean_series) > 0:
        skew_val = clean_series.skew()
        skewness_records.append({
            'Feature': col,
            'Skewness': round(skew_val, 2),
            'Skew_Direction': 'Right (Positive)' if skew_val > 0.5 else ('Left(Negative)' if skew_val < 0.5 else 'Symmetric')
            
        })
skew_train_df = pd.DataFrame(skewness_records).sort_values (by = 'Skewness', ascending = False, key = abs).reset_index(drop = True)
print("--- Top 15 Most Skewed Features in Train Dataset ---")

display(skew_train_df)
```

**Outliers Inspection & Distribution Skewness Analysis**
- Impact of the previous stage: Generated engineered temporal and encoded categorical features. 
- Stage: Data Preprocessing - Skewness Analysis
- Requirement: Calculate skewness coefficients for all numeric features in the training dataset. 
- Intention: Identify feature distribution asymmetries to guide preprocessing and scaling strategies. 
- Solution: Extracted numeric columns, computed statistical skewness values, and ranked features in descending order.

```python
# ============================================================================================
# 2.7 : Data Preprocessing | Outlier Detection & Diagnostics (IQR Method on Updated Datasets)
# ============================================================================================
print("=" * 80)
print("2.7 OUTLIER DETECTION & DIAGNOSTICS (IQR METHOD)")
print("=" * 80)

def compute_iqr_outliers(df, feature_list, dataset_name="Dataset"):
    outlier_records = []
    for col in feature_list:
        clean_series = pd.to_numeric(df[col], errors='coerce').dropna()
        if len(clean_series) == 0:
            continue
        q1 = np.percentile(clean_series, 25)
        q3 = np.percentile(clean_series, 75)
        iqr = q3 - q1
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr
        
        outlier_count = ((clean_series < lower_bound) | (clean_series > upper_bound)).sum()
        outlier_pct = (outlier_count / len(df)) * 100
        
        if outlier_count > 0:
            outlier_records.append({
                'Feature': col,
                'Outlier_Count': outlier_count,
                'Outlier_%': round(outlier_pct, 2),
                'Lower_Bound': round(lower_bound, 4),
                'Upper_Bound': round(upper_bound, 4)
            })
            
    return pd.DataFrame(outlier_records).sort_values(by='Outlier_Count', ascending=False).reset_index(drop=True)

outliers_train = compute_iqr_outliers(train_df, train_df_num, "Train Set")
print("--- Top 10 Features with Highest Outlier Volume (Train) ---")
display(outliers_train)
```

```python
# ==============================================================================
# 2.8 : Feature Scaling & Time-Based Validation Splitting (Safe Print)
# ==============================================================================

from sklearn.preprocessing import StandardScaler
# import pandas as pd
# import numpy as np

# Ensure Date is strictly datetime
train_df['Date'] = pd.to_datetime(train_df['Date'], errors='coerce')

# 1. Define feature columns (clean ASCII list)
feature_cols = [
    'Store', 'DayOfWeek', 'Customers', 'Open', 'Promo', 'SchoolHoliday', 
    'Year', 'Month', 'Day', 'WeekOfYear'
]

state_holiday_cols = [col for col in train_df.columns if col.startswith('StateHoliday_')]
feature_cols.extend(state_holiday_cols)

X = train_df[feature_cols]
y = train_df['Sales']

# 2. Feature Scaling for Continuous Variables (Customers)
scaler = StandardScaler()
X_scaled = X.copy()
X_scaled['Customers'] = scaler.fit_transform(X[['Customers']])

print("[Preprocessing] StandardScaler applied successfully to 'Customers'.")
print(f"Scaled Customers Mean: {X_scaled['Customers'].mean():.2f} | Std: {X_scaled['Customers'].std():.2f}")

# 3. Time-Based Validation Split (Chronological Split)
train_df_sorted = train_df.sort_values('Date').reset_index(drop=True)

split_index = int(len(train_df_sorted) * 0.8)

# Construct chronologically ordered train and validation sets
X_train_time = train_df_sorted.loc[:split_index, feature_cols].copy()
y_train_time = train_df_sorted.loc[:split_index, 'Sales'].copy()

X_val_time = train_df_sorted.loc[split_index+1:, feature_cols].copy()
y_val_time = train_df_sorted.loc[split_index+1:, 'Sales'].copy()

# Apply scaling parameters to time-based split
X_train_time['Customers'] = scaler.fit_transform(X_train_time[['Customers']])
X_val_time['Customers'] = scaler.transform(X_val_time[['Customers']])

# Safe Date Bounds Extraction
min_date = train_df_sorted['Date'].min()
split_date = train_df_sorted.loc[split_index, 'Date']
val_start_date = train_df_sorted.loc[split_index+1, 'Date']
max_date = train_df_sorted['Date'].max()

print(f"\n[Validation Split] Chronological Split Complete:")
print(f"Training Period Records: {len(X_train_time)} (From {min_date.date() if pd.notnull(min_date) else 'N/A'} to {split_date.date() if pd.notnull(split_date) else 'N/A'})")
print(f"Validation Period Records: {len(X_val_time)} (From {val_start_date.date() if pd.notnull(val_start_date) else 'N/A'} to {max_date.date() if pd.notnull(max_date) else 'N/A'})")
```
